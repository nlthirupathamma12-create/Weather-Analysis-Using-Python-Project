```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np



```

```python
# Load dataset
df=pd.read_csv(r"C:\Users\user\OneDrive\Documents\Weather_Data.csv")
```

```python
# Ensure the 'Date/Time' column is parsed as datetime
df['Date/Time'] = pd.to_datetime(df['Date/Time'])
```

```python
plt.figure(figsize=(14, 7))
plt.plot(df['Date/Time'], df['Temp_C'], label='Temperature (C)', color='black')
plt.xlabel('Date/Time')
plt.ylabel('Temperature (C)')
plt.title('Temperature Over Time')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Wind Speed_km/h'], bins=30, color='blue', edgecolor='black')
plt.xlabel('Wind Speed (km/h)')
plt.ylabel('Frequency')
plt.title('Distribution of Wind Speeds')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
correlation_matrix = df[['Temp_C', 'Dew Point Temp_C', 'Rel Hum_%', 'Wind Speed_km/h', 'Visibility_km', 'Press_kPa']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
np.random.seed(0)
df = pd.DataFrame({
    'Weather Condition': np.random.choice(['Sunny', 'Cloudy', 'Rainy', 'Windy'], 100),
    'Wind Speed_km/h': np.random.normal(15, 5, 100)
})
```

```python
plt.figure(figsize=(12, 6))
palette = {
   'Sunny': 'blue',
    'Windy': 'red',
    'Cloudy': 'orange',
    'Rainy': 'green'
}
sns.boxplot(x='Weather Condition', y='Wind Speed_km/h',hue='Weather Condition',data=df,palette=palette,legend=False)
plt.xlabel('Weather Condition')
plt.ylabel('Wind Speed (km/h)')
plt.title('Box Plot of Wind Speed by Weather Condition')
plt.show()
```

```python
data = {
    'Weather Condition': ['Sunny', 'Cloudy', 'Rainy', 'Windy'],
    'Frequency': [25, 30, 20, 25]
}
colors={
    'Sunny':'blue',
    'Cloudy':'orange',
    'Rainy':'green',
    'Windy':'red'
}
df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(8, 8))
plt.pie(df['Frequency'], labels=df['Weather Condition'], autopct='%1.1f%%', startangle=140,colors=[colors[cond] for cond in df['Weather Condition']])
plt.title('Weather Condition Distribution')
plt.show()
```

```python
plt.figure(figsize=(14, 7))
palette = {
   'Sunny': 'blue',
    'Windy': 'red',
    'Cloudy': 'orange',
    'Rainy': 'green'
}
sns.countplot(data=df, y='Weather Condition', order=df['Weather Condition'].value_counts().index,hue='Weather Condition',palette=palette,legend=False)
plt.xlabel('Count')
plt.ylabel('Weather Condition')
plt.title('Count of Each Weather Condition')
plt.show()
```
